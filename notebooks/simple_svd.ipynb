# SVD Demonstration with Orthogonal Vectors
## Exploring Principal Components of Noisy Linear Data

### 1. Theoretical Background
Given:
- Vector $\mathbf{v}$ (norm 1) defining primary direction
- Orthogonal vector $\mathbf{u}$ (norm 1)
- Points $x_i = a_i \mathbf{v} + b_i \mathbf{u}$ with small $b_i$

The SVD of matrix $\mathbf{X}$ (containing points as rows) should reveal:
- First right singular vector ($\mathbf{V}$[:, 0]) aligns with $\mathbf{v}$
- Second right singular vector aligns with $\mathbf{u}$

````python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
````

### 2. Create Synthetic Data
Generate points along v = [1, 0] with small orthogonal noise:

````python
# Base vectors
v = np.array([1, 0])
u = np.array([0, 1])

# Generate points
n = 100
a = np.linspace(-5, 5, n)
b = 0.2 * np.random.randn(n)  # Small noise

X = np.outer(a, v) + np.outer(b, u)
````

### 3. Compute SVD
Perform Singular Value Decomposition and analyze components:

````python
U, S, Vt = np.linalg.svd(X, full_matrices=False)
V = Vt.T

print("Right singular vectors:\n", V)
print("\nSingular values:", S)
````

### 4. Visualization
Plot the data and principal components:

````python
plt.figure(figsize=(10,6))

# Plot points
plt.scatter(X[:,0], X[:,1], alpha=0.5, label='Data points')

# Plot true direction
plt.quiver(0, 0, v[0], v[1], 
          scale=5, color='r', 
          label='True direction (v)')

# Plot SVD direction
plt.quiver(0, 0, V[0,0], V[1,0],
          scale=5, color='g', linestyle='--',
          label='SVD direction (V[:,0])')

plt.axis('equal')
plt.title('Data Distribution and Principal Components')
plt.legend()
plt.show()
````

### 5. Analysis
Key observations:
- First singular vector (green) aligns with true direction `v` (red)
- First singular value is much larger than second
- Data variance is mostly explained along `v` direction

Numerical verification:
```python
np.allclose(V[:,0], v)  # Should return True
S[0] > S[1]  # Should return True
```

### 6. Extended Example: Rotated Basis
Same concept works for arbitrary orthogonal bases:

````python
theta = np.pi/6  # 30 degree rotation
v_rot = np.array([np.cos(theta), np.sin(theta)])
u_rot = np.array([-np.sin(theta), np.cos(theta)])

# Generate rotated data
X_rot = np.outer(a, v_rot) + np.outer(0.2*np.random.randn(n), u_rot)

# Compute SVD
_, _, Vt_rot = np.linalg.svd(X_rot)
V_rot = Vt_rot.T

# Plot rotated version
plt.scatter(X_rot[:,0], X_rot[:,1], alpha=0.5)
plt.quiver(0, 0, v_rot[0], v_rot[1], scale=5, color='r')
plt.quiver(0, 0, V_rot[0,0], V_rot[1,0], scale=5, color='g', linestyle='--')
plt.axis('equal')
plt.title('Rotated Basis Demonstration')
plt.show()
````

### 7. Conclusion
The SVD successfully recovers:
- True direction `v` as first principal component
- Orthogonal direction `u` as second component
- Demonstrated robustness through coordinate rotation

Key takeaway: SVD identifies directions of maximum variance, which correspond to the underlying data structure even with added noise.
